## Explication du résumé technique

- Il y a 505 lignes (clients) dans le fichier.
- "non-null" veut dire qu'il n'y a pas de case vide pour cette colonne.
- Les types de données sont : object (texte), int64 (nombres entiers), float64 (nombres à virgule).

```python
print(df["Age"].mean(), df["Age"].median())
```

```python
print(df["Customer spendings"].mean(), df["Customer spendings"].median())
```

```python
df.groupby("Country")["Age"].median()
```

```python
depenses_par_pays = df.groupby("Country")["Customer spendings"].sum()
depenses_par_pays.plot(kind="bar", figsize=(10,5))
plt.title("Dépenses des clients par pays")
plt.xlabel("Pays")
plt.ylabel("Dépenses totales")
plt.show()
```

```python
df = df[df["Customer spendings"] >= 10]
```

```python
df = df.drop_duplicates()
```

```python
df_final = df[["Country", "Age", "Gender", "Customer spendings"]]
df_final = df_final.rename(columns={"Age": "Âge"})
df_final.to_csv("dataset-sell4all-clean.csv", index=False)
```
